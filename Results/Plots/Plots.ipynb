```python
import os
import numpy as np
import matplotlib.pyplot as plt
from sklearn.manifold import TSNE

# ============================================================
# ROOT DIRECTORY
# ============================================================

ROOT_DIR = "/content/drive/MyDrive/IIITH_Voice/TESS_Time_Embeddings_hubert"

# ============================================================
# GET SUBFOLDERS
# ============================================================

emotion_folders = sorted([
    folder for folder in os.listdir(ROOT_DIR)
    if os.path.isdir(os.path.join(ROOT_DIR, folder))
])

print("Detected Folders:")
print(emotion_folders)

# ============================================================
# LOAD EMBEDDINGS
# ============================================================

all_embeddings = []
all_labels = []

max_len = 0

# ------------------------------------------------------------
# FIRST PASS → FIND MAX TIME LENGTH
# ------------------------------------------------------------

for folder_name in emotion_folders:

    folder_path = os.path.join(ROOT_DIR, folder_name)

    npy_files = [
        f for f in os.listdir(folder_path)
        if f.endswith(".npy")
    ]

    for file in npy_files:

        file_path = os.path.join(folder_path, file)

        embedding = np.load(file_path)

        if len(embedding.shape) > 1:

            max_len = max(max_len, embedding.shape[0])

print("\nMaximum Sequence Length:", max_len)

# ------------------------------------------------------------
# SECOND PASS → PAD EMBEDDINGS
# ------------------------------------------------------------

for label_idx, folder_name in enumerate(emotion_folders):

    folder_path = os.path.join(ROOT_DIR, folder_name)

    npy_files = [
        f for f in os.listdir(folder_path)
        if f.endswith(".npy")
    ]

    print(f"{folder_name}: {len(npy_files)} files")

    for file in npy_files:

        file_path = os.path.join(folder_path, file)

        embedding = np.load(file_path)

        # ====================================================
        # PAD SEQUENCES
        # ====================================================

        if len(embedding.shape) > 1:

            time_steps, feat_dim = embedding.shape

            padded = np.zeros((max_len, feat_dim))

            padded[:time_steps] = embedding

            embedding = padded

        # Flatten to 1D vector
        embedding = embedding.flatten()

        all_embeddings.append(embedding)

        all_labels.append(label_idx)

# ============================================================
# CONVERT TO ARRAYS
# ============================================================

all_embeddings = np.vstack(all_embeddings)

all_labels = np.array(all_labels)

print("\nEmbeddings Shape:", all_embeddings.shape)
print("Labels Shape:", all_labels.shape)

# ============================================================
# t-SNE
# ============================================================

tsne = TSNE(
    n_components=2,
    perplexity=30,
    random_state=42
)

embeddings_2d = tsne.fit_transform(all_embeddings)

# ============================================================
# PLOT
# ============================================================

plt.figure(figsize=(12,8))

for i, folder_name in enumerate(emotion_folders):

    idx = all_labels == i

    plt.scatter(
        embeddings_2d[idx, 0],
        embeddings_2d[idx, 1],
        label=folder_name,
        alpha=0.6
    )

plt.title("t-SNE Visualization of HuBERT Temporal Embeddings")

plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")

plt.legend(fontsize=8)

plt.tight_layout()

# ============================================================
# SAVE
# ============================================================

plt.savefig(
    "/content/drive/MyDrive/IIITH_Voice/voice_tsne_hubert.png",
    dpi=300
)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.manifold import TSNE
from tensorflow.keras.models import Model
import tensorflow as tf
import pickle
import pandas as pd
from tensorflow.keras.preprocessing.sequence import pad_sequences

# ============================================================
# CONFIG
# ============================================================

SAVE_DIR = "/content/drive/MyDrive/IIITH_Text"

MODEL_PATH = f"{SAVE_DIR}/final_text_model.keras"
TOKENIZER_PATH = f"{SAVE_DIR}/tokenizer.pkl"
LABEL_ENCODER_PATH = f"{SAVE_DIR}/label_encoder.pkl"
TEST_SPLIT_PATH = f"{SAVE_DIR}/text_test_split.csv"

MAX_TOKENS = 8

# ============================================================
# LOAD MODEL
# ============================================================

model = tf.keras.models.load_model(MODEL_PATH)

# ============================================================
# LOAD TOKENIZER
# ============================================================

with open(TOKENIZER_PATH, "rb") as f:
    tokenizer = pickle.load(f)

# ============================================================
# LOAD LABEL ENCODER
# ============================================================

with open(LABEL_ENCODER_PATH, "rb") as f:
    label_encoder = pickle.load(f)

# ============================================================
# LOAD TEST DATA
# ============================================================

test_df = pd.read_csv(TEST_SPLIT_PATH)

texts = test_df["transcription"].astype(str)

labels = test_df["y"].values

# ============================================================
# TEXT → SEQUENCES
# ============================================================

X_test = tokenizer.texts_to_sequences(texts)

X_test = pad_sequences(
    X_test,
    maxlen=MAX_TOKENS,
    padding="post"
)

# ============================================================
# CREATE EMBEDDING MODEL
# ============================================================

embedding_model = Model(
    inputs=model.input,
    outputs=model.get_layer("text_representation").output
)

# ============================================================
# EXTRACT EMBEDDINGS
# ============================================================

text_embeddings = embedding_model.predict(X_test)

print("Embeddings Shape:", text_embeddings.shape)

# ============================================================
# t-SNE
# ============================================================

tsne = TSNE(
    n_components=2,
    perplexity=30,
    random_state=42
)

text_2d = tsne.fit_transform(text_embeddings)

# ============================================================
# PLOT
# ============================================================

emotion_names = label_encoder.classes_

plt.figure(figsize=(10,8))

for i, emotion in enumerate(emotion_names):

    idx = labels == i

    plt.scatter(
        text_2d[idx, 0],
        text_2d[idx, 1],
        label=emotion,
        alpha=0.7
    )

plt.title("t-SNE Visualization of Text Emotion Embeddings")

plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")

plt.legend()

plt.tight_layout()

# ============================================================
# SAVE
# ============================================================

plt.savefig(
    f"{SAVE_DIR}/text_tsne.png",
    dpi=300
)

plt.show()
```

```python
import torch
import torch.nn as nn
import numpy as np
import matplotlib.pyplot as plt

from tqdm import tqdm
from sklearn.manifold import TSNE
from transformers import AutoTokenizer, AutoModel
from torch.utils.data import Dataset, DataLoader
from torch.nn.utils.rnn import pad_sequence
import pandas as pd
import os

# ============================================================
# CONFIG
# ============================================================

CSV_PATH = "/content/drive/MyDrive/IIITH_Text/tess_full_metadata.csv"

EMBEDDING_DIR = "/content/drive/MyDrive/IIITH_Voice/TESS_Time_Embeddings_hubert"

MODEL_PATH = "/content/drive/MyDrive/IIITH_Text/multimodal_emotion_model.pt"

TEXT_MODEL = "bert-base-uncased"

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

MAX_LEN = 32
BATCH_SIZE = 8

# ============================================================
# LOAD DATAFRAME
# ============================================================

df = pd.read_csv(CSV_PATH)

# ============================================================
# FIX SPEAKER NAMES
# ============================================================

df["speaker"] = df["speaker"].replace(
    {"OA": "OAF"}
)

# ============================================================
# LABELS
# ============================================================

emotion_names = sorted(df["emotion"].unique())

emotion_to_id = {
    emotion: idx
    for idx, emotion in enumerate(emotion_names)
}

df["label_id"] = df["emotion"].map(emotion_to_id)

# ============================================================
# TOKENIZER
# ============================================================

tokenizer = AutoTokenizer.from_pretrained(
    TEXT_MODEL
)

# ============================================================
# DATASET
# ============================================================

class EmotionDataset(Dataset):

    def __init__(self, dataframe):

        self.df = dataframe.reset_index(drop=True)

    def __len__(self):

        return len(self.df)

    def __getitem__(self, idx):

        row = self.df.iloc[idx]

        filename = row["filename"]

        text = row["transcription"]

        label = row["label_id"]

        base = os.path.splitext(filename)[0]

        # ====================================================
        # FIND MATCHING FOLDER
        # ====================================================

        emotion_name = row["emotion"].lower()

        # Handle pleasant surprise
        if emotion_name == "ps":
            emotion_name = "pleasant"

        matching_folder = None

        for folder in os.listdir(EMBEDDING_DIR):

            if (
                row["speaker"] in folder
                and emotion_name in folder.lower()
            ):
                matching_folder = folder
                break

        if matching_folder is None:

            raise FileNotFoundError(
                f"No folder found for "
                f"{row['speaker']} {row['emotion']}"
            )

        # ====================================================
        # BUILD EMBEDDING PATH
        # ====================================================

        emb_path = os.path.join(
            EMBEDDING_DIR,
            matching_folder,
            f"{base}.npy"
        )

        # ====================================================
        # CHECK FILE EXISTS
        # ====================================================

        if not os.path.exists(emb_path):

            raise FileNotFoundError(
                f"Embedding file missing:\n{emb_path}"
            )

        # ====================================================
        # LOAD EMBEDDING
        # ====================================================

        speech_embedding = np.load(emb_path)

        speech_embedding = torch.tensor(
            speech_embedding,
            dtype=torch.float
        )

        # ====================================================
        # TOKENIZE TEXT
        # ====================================================

        encoding = tokenizer(
            text,
            truncation=True,
            padding="max_length",
            max_length=MAX_LEN,
            return_tensors="pt"
        )

        return {
            "speech": speech_embedding,
            "input_ids": encoding["input_ids"].squeeze(0),
            "attention_mask": encoding["attention_mask"].squeeze(0),
            "label": torch.tensor(label)
        }

# ============================================================
# COLLATE FUNCTION
# ============================================================

def collate_fn(batch):

    speech = [
        item["speech"]
        for item in batch
    ]

    speech = pad_sequence(
        speech,
        batch_first=True,
        padding_value=0
    )

    input_ids = torch.stack([
        item["input_ids"]
        for item in batch
    ])

    attention_mask = torch.stack([
        item["attention_mask"]
        for item in batch
    ])

    labels = torch.stack([
        item["label"]
        for item in batch
    ])

    return {
        "speech": speech,
        "input_ids": input_ids,
        "attention_mask": attention_mask,
        "labels": labels
    }

# ============================================================
# DATALOADER
# ============================================================

dataset = EmotionDataset(df)

loader = DataLoader(
    dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    collate_fn=collate_fn
)

# ============================================================
# ATTENTION POOLING
# ============================================================

class AttentionPooling(nn.Module):

    def __init__(self, hidden_dim):

        super().__init__()

        self.attention = nn.Linear(
            hidden_dim,
            1
        )

    def forward(self, x):

        weights = torch.softmax(
            self.attention(x),
            dim=1
        )

        pooled = torch.sum(
            weights * x,
            dim=1
        )

        return pooled

# ============================================================
# MODEL
# ============================================================

class MultimodalEmotionModel(nn.Module):

    def __init__(self, num_classes):

        super().__init__()

        # ====================================================
        # SPEECH BRANCH
        # ====================================================

        self.speech_lstm = nn.LSTM(
            input_size=768,
            hidden_size=256,
            batch_first=True,
            bidirectional=True,
            num_layers=2,
            dropout=0.3
        )

        self.speech_attention = AttentionPooling(
            512
        )

        self.speech_fc = nn.Linear(
            512,
            256
        )

        # ====================================================
        # TEXT BRANCH
        # ====================================================

        self.text_encoder = AutoModel.from_pretrained(
            TEXT_MODEL
        )

        self.text_fc = nn.Linear(
            768,
            256
        )

        # ====================================================
        # CROSS ATTENTION
        # ====================================================

        self.cross_attention = nn.MultiheadAttention(
            embed_dim=256,
            num_heads=4,
            batch_first=True
        )

        # ====================================================
        # CLASSIFIER
        # ====================================================

        self.classifier = nn.Sequential(

            nn.Linear(512, 256),

            nn.ReLU(),

            nn.Dropout(0.3),

            nn.Linear(256, num_classes)
        )

# ============================================================
# LOAD MODEL
# ============================================================

checkpoint = torch.load(
    MODEL_PATH,
    map_location=DEVICE,
    weights_only=False
)

model = MultimodalEmotionModel(
    num_classes=len(emotion_names)
)

model.load_state_dict(
    checkpoint["model_state_dict"]
)

model.to(DEVICE)

model.eval()

# ============================================================
# EXTRACT FUSION EMBEDDINGS
# ============================================================

all_embeddings = []
all_labels = []

with torch.no_grad():

    for batch in tqdm(loader):

        speech = batch["speech"].to(DEVICE)

        input_ids = batch["input_ids"].to(DEVICE)

        attention_mask = batch[
            "attention_mask"
        ].to(DEVICE)

        labels = batch["labels"].to(DEVICE)

        # ====================================================
        # SPEECH BRANCH
        # ====================================================

        speech_out, _ = model.speech_lstm(
            speech
        )

        speech_repr = model.speech_attention(
            speech_out
        )

        speech_repr = model.speech_fc(
            speech_repr
        )

        # ====================================================
        # TEXT BRANCH
        # ====================================================

        text_outputs = model.text_encoder(
            input_ids=input_ids,
            attention_mask=attention_mask
        )

        text_repr = text_outputs.last_hidden_state[:,0]

        text_repr = model.text_fc(
            text_repr
        )

        # ====================================================
        # WEIGHTED MODALITIES
        # ====================================================

        speech_repr = speech_repr * 0.7

        text_repr = text_repr * 0.3

        # ====================================================
        # CROSS ATTENTION
        # ====================================================

        speech_q = speech_repr.unsqueeze(1)

        text_kv = text_repr.unsqueeze(1)

        attended, _ = model.cross_attention(
            query=speech_q,
            key=text_kv,
            value=text_kv
        )

        attended = attended.squeeze(1)

        # ====================================================
        # FUSION VECTOR
        # ====================================================

        fused = torch.cat([
            speech_repr,
            attended
        ], dim=1)

        all_embeddings.append(
            fused.cpu().numpy()
        )

        all_labels.append(
            labels.cpu().numpy()
        )

# ============================================================
# CONCATENATE
# ============================================================

fusion_embeddings = np.concatenate(
    all_embeddings,
    axis=0
)

fusion_labels = np.concatenate(
    all_labels,
    axis=0
)

print("Fusion Embeddings Shape:", fusion_embeddings.shape)
print("Fusion Labels Shape:", fusion_labels.shape)

# ============================================================
# t-SNE
# ============================================================

tsne = TSNE(
    n_components=2,
    perplexity=30,
    random_state=42
)

fusion_2d = tsne.fit_transform(
    fusion_embeddings
)

# ============================================================
# PLOT
# ============================================================

plt.figure(figsize=(12,8))

for i, emotion in enumerate(emotion_names):

    idx = fusion_labels == i

    plt.scatter(
        fusion_2d[idx, 0],
        fusion_2d[idx, 1],
        label=emotion,
        alpha=0.7
    )

plt.title("t-SNE Visualization of Fusion Emotion Embeddings")

plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")

plt.legend()

plt.tight_layout()

# ============================================================
# SAVE FIGURE
# ============================================================

plt.savefig(
    "/content/drive/MyDrive/IIITH_Text/fusion_tsne.png",
    dpi=300
)

plt.show()
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix

# Class labels
classes = [
    'angry',
    'disgust',
    'fear',
    'happy',
    'neutral',
    'pleasant_surprise',
    'sad'
]

# Example true labels and predicted labels
# Since your accuracy is 100%, predictions = true labels
y_true = []
y_pred = []

for i in range(len(classes)):
    y_true.extend([i] * 80)
    y_pred.extend([i] * 80)

# Generate confusion matrix
cm = confusion_matrix(y_true, y_pred)

# Plot heatmap
plt.figure(figsize=(10, 7))
sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=classes,
    yticklabels=classes
)

plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix Heatmap')

plt.tight_layout()
plt.show()
```
