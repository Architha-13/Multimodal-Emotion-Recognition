```python
import os
import pickle
import numpy as np
import tensorflow as tf

from tensorflow.keras.preprocessing.sequence import pad_sequences

SAVE_DIR = "/content/drive/MyDrive/IIITH_Text"

MODEL_PATH = f"{SAVE_DIR}/final_text_model.keras"
TOKENIZER_PATH = f"{SAVE_DIR}/tokenizer.pkl"
LABEL_ENCODER_PATH = f"{SAVE_DIR}/label_encoder.pkl"

MAX_TOKENS = 8


print("Loading model...")

model = tf.keras.models.load_model(MODEL_PATH)


with open(TOKENIZER_PATH, "rb") as f:
    tokenizer = pickle.load(f)


with open(LABEL_ENCODER_PATH, "rb") as f:
    label_encoder = pickle.load(f)

print("Model Loaded Successfully!")


def predict_emotion(text):


    sequence = tokenizer.texts_to_sequences([text])

    padded = pad_sequences(
        sequence,
        maxlen=MAX_TOKENS,
        padding="post"
    )

    prediction = model.predict(padded)

    pred_index = np.argmax(prediction)

    pred_emotion = label_encoder.inverse_transform([pred_index])[0]

    confidence = prediction[0][pred_index]

    
    print("\n===================================")
    print("Input Text :", text)
    print("Predicted Emotion :", pred_emotion)
    print("Confidence :", round(float(confidence) * 100, 2), "%")
    print("===================================\n")


    print("All Emotion Scores:\n")

    for emotion, score in zip(label_encoder.classes_, prediction[0]):

        print(f"{emotion:10s} : {score:.4f}")

predict_emotion("say the word bath")
```
